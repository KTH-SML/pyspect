# Exact symbolic reachability for integer signals

Two signals range over $\mathbb{N}_0$, may never agree, and at most one changes per event. We compute `safe UNTIL goal` exactly—without a grid or state-space enumeration.

> Install the optional backend with `pip install -e ".[symbolic]"`.

```python
import portion as P

from pyspect import ExactDiscLTL, TLT, UNTIL
from pyspect.impls.symbolic import (
    AllDifferent, OneVariableTransition, SymbolicSet, SymbolicSystem,
)

system = SymbolicSystem(
    variables=("j1", "j2"),
    invariant=AllDifferent(),
    transition=OneVariableTransition(),
)
```

```python
winning = TLT(
    UNTIL("safe", "goal"),
    primitives=ExactDiscLTL,
    where={
        "safe": SymbolicSet(j1=P.closed(0, 1)),
        "goal": SymbolicSet(j1=2, j2=1),
    },
).realize(system)

winning
```

```python
samples = [(0, 3), (1, 0), (2, 1), (2, 3), (3, 4)]
{state: state in winning for state in samples}
```

## Fixed-point lattice view

The synthetic `k` axis shows the cumulative approximants already computed by `reach()`: `k=0` is the goal and the last column is the fixed point. It is an iteration index, not physical time. The state window and projected hidden variables are inferred automatically.

```python
goal = system.states(j1=2, j2=1)
system.plot(
    (winning, {"name": "winning", "color": "#4C78A8"}),
    (goal, {"name": "goal", "color": "#F58518"}),
    axes=("k", "j1"),
    layout_title="Growth of the safe UNTIL goal winning set",
)
```
